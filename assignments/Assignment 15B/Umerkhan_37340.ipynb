```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("~/Documents/Assignment 15B/student_tea_snacks _1_.csv")
df.head()
```

```python
# Q1: What is the most preferred tea type?
tea = df["tea_type"].value_counts()
print(tea)
print("The most preferred tea type is", tea.idxmax(), "with", tea.max(), "students.")
```

```python
# Q2: What is the average number of cups per gender?
average_cups = df.groupby("gender")["cups_per_day"].mean()
print(average_cups.round(2))
print("Males drink", round(average_cups["Male"], 2), "cups per day on average.")
print("Females drink", round(average_cups["Female"], 2), "cups per day on average.")
print("The gender that drinks more cups on average is", average_cups.idxmax())
```

```python
# Q3: Make a bar chart of snack choices.
snacks = df["snack_choice"].value_counts()
print(snacks)
snacks.plot(kind="bar")
plt.title("Student Snack Choices")
plt.xlabel("Snack")
plt.ylabel("Number of Students")
plt.show()
print("The most popular snack is", snacks.idxmax(), "with", snacks.max(), "students.")
```

```python
# Q4: Compare tea preference by age group.
tea_age = pd.crosstab(df["age_group"], df["tea_type"])
print(tea_age)
for age in tea_age.index:
    print("The most preferred tea for age group", age, "is", tea_age.loc[age].idxmax())
```

```python
# Q5: Count the number of students who don't drink tea.
tea_preference = df["prefers_tea"].value_counts()
print(tea_preference)
no_tea = tea_preference["No"]
print("The number of students who do not prefer tea is", no_tea)
```
