```python
# DATASET 1
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("~/Documents/assignment 15A/pakistani_ecommerce_dataset.csv")
df.head()
```

```python
# QUESTION 1: MOST POPULAR PRODUCT CATEGORY
category= df["product_category"].value_counts()
print(category)
print("The most popular product category is", category.idxmax(), "with", category.max(), "orders.")

```

```python
# Q2 — Average order value
average = df["order_value"].mean()
print(("The aversge order value is"),round(average, 2))


```

```python
# Q3 — City with the most orders
city = df["city"].value_counts()
print(city)
print("The city with the highest number of orders is", city.idxmax(), "with", city.max(), "orders.")


```

```python
# Q4. What is the most used payment method?
payment = df["payment_method"].value_counts()
print(payment)
print("The most used payment method is", payment.idxmax(), "with", payment.max(), "orders.")
```

```python
# Q5. Draw a bar chart showing average order value by city.( USED CHATGPT FOR THIS)
city_average = df.groupby("city")["order_value"].mean()
print(city_average.round(2))
city_average.plot(kind="bar")
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value (PKR)")
plt.show()
print("The city with the highest average order value is", city_average.idxmax(), "with PKR", round(city_average.max(), 2))
```

```python
# DATASET 2: student_lifestyle_dataset.csv
df20 =pd.read_csv("~/Documents/assignment 15A/student_lifestyle_dataset.csv")
df20.head()
```

```python
# Q1. What is the average number of study hours?
average_study = df20["study_hours"].mean()
print("The average number of study hours is", round(average_study, 2), "hours.")
```

```python
# Q2. Who sleeps more on average males or females?
average_sleep = df20.groupby("gender")["sleep_hours"].mean()
average_sleep.round(2)
print("The gender that sleeps more on average is", average_sleep.idxmax(), "with", round(average_sleep.max(), 2), "hours.")
```

```python
# Q3. Is there a relationship between screen time and sleep hours?
correlation = df20["screen_time"].corr(df20["sleep_hours"])
print("The correlation between screen time and sleep hours is", round(correlation, 2))
if abs(correlation) < 0.1:
    print("There is almost no linear relationship between screen time and sleep hours.")
elif correlation > 0:
    print("There is a positive relationship between screen time and sleep hours.")
else:
    print("There is a negative relationship between screen time and sleep hours.")
```

```python
# Q4. Draw a histogram of daily transport cost.(USED CHAT GPT FOR THIS)
plt.hist(df20["daily_transport_cost"])
plt.title("Daily Transport Cost")
plt.xlabel("Transport Cost (PKR)")
plt.ylabel("Number of Students")
plt.show()
average_transport = df20["daily_transport_cost"].mean()
print("The average daily transport cost is PKR", round(average_transport, 2))
```

```python
# Q5. Compare average internet usage by gender.
average_internet = df20.groupby("gender")["internet_usage"].mean()
print(average_internet.round(2))
print("The gender with higher average internet usage is", average_internet.idxmax(), "with", round(average_internet.max(), 2))
```
