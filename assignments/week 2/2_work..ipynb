```python
import pandas as pd
```

```python
df = pd.read_csv("~/documents/prices-desk-assignment/week 2/teaching_prices_messy.csv")
```

```python
print(df.head())

print("Rows and columns:", df.shape)
```

```python
print(df.isna().sum())
print(df.duplicated().sum())
print(df["item"].value_counts())
print(df.dtypes)
```

```python
prices = pd.to_numeric(
    df["price"].str.replace(",",""),errors="coerce")
print(df[prices>500])
```

```python
clean = df.copy()
clean["prices"] = prices
print(clean.head())
```

```python
clean["price"] = pd.to_numeric(
    clean["price"].str.replace(",", ""),
    errors="coerce"
).astype("float64")
clean.loc[clean["price"] > 500, "price"] = float("nan")
print(clean["price"].isna().sum())
```

```python
print(clean[clean["item"] == "oil b"])
clean.loc[clean["item"] == "oil b", "item"] = "Staple A"
print(clean["item"].value_counts())
```

```python
clean = clean.drop_duplicates()
print(clean.duplicated().sum())
print(clean.shape)
```

```python
clean = clean.drop(columns=["prices"])
print(clean.shape)
```

```python

print("BEFORE CLEANING")
print("Rows:", len(df))
print("Missing prices:", df["price"].isna().sum())
print("Duplicates:", df.duplicated().sum())

print("AFTER CLEANING")
print("Rows:", len(clean))
print("Missing prices:", clean["price"].isna().sum())
print("Duplicates:", clean.duplicated().sum())
```
