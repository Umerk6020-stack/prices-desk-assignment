```python
import pandas as pd
df= pd.read_csv("~/Documents/prices-desk-assignment/Week 3/teaching_prices_clean-2.csv")
print(df.head())
print(df.shape)
```

```python
print(df.groupby("item")["price"].mean().round(2))
print(df.groupby("item")["price"].median().round(2))
```

```python
for item in df["item"].unique():
    data = df[df["item"] == item]

    first = data[data["date"] == data["date"].min()]["price"].mean()
    last = data[data["date"] == data["date"].max()]["price"].mean()

    change = (last - first) / first * 100

    print(item, round(change, 2), "%")
```

```python

# Select Staple A
staple = df[df["item"] == "Staple A"].copy()

# Without the suspicious price
print("WITHOUT 1250")
print("Mean:", round(staple["price"].mean(), 2))
print("Median:", staple["price"].median())

# Put the suspicious price back for comparison
with_flag = staple.copy()

with_flag.loc[
    (with_flag["date"] == "2026-01-15") &
    (with_flag["city"] == "City 2"),
    "price"
] = 1250

# With the suspicious price
print("WITH 1250")
print("Mean:", round(with_flag["price"].mean(), 2))
print("Median:", with_flag["price"].median())
```
