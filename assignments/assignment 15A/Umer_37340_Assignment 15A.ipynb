```python
import pandas as pd
df = pd.read_csv("ride_hailing_dataset.csv")
df.head()
```

```python
# Q1: What is the average fare by service type?
average_fare=df.groupby("service_type")["fare"].mean().round(2)
print(average_fare.to_string())
print("The service type with the highest average fare is",average_fare.idxmax(),"with Rs.",average_fare.max())
```

```python
# Q2: Which city has the most trips?
city_trips=df["city"].value_counts()
print(city_trips.to_string())
print("The city with the most trips is",city_trips.idxmax(),"with",city_trips.max(),"trips.")
```

```python
# Q3: What is the average distance per trip?
average_distance=df["distance_km"].mean()
print("The average distance per trip is",round(average_distance,2),"km.")
```

```python
# Q4: Compare trip ratings by gender.
gender_ratings=df.groupby("gender")["trip_rating"].mean().round(2)
print(gender_ratings.to_string())
print("The gender with the higher average trip rating is",gender_ratings.idxmax(),"with",gender_ratings.max())
```

```python
# Q5: Make a bar chart of service type vs. average fare.
import matplotlib.pyplot as plt
average_fare=df.groupby("service_type")["fare"].mean()
average_fare.plot(kind="bar",color="skyblue")
plt.title("Average Fare by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Average Fare (Rs.)")
plt.xticks(rotation=0)
plt.show()
print("InDrive has the highest average fare.")
```
